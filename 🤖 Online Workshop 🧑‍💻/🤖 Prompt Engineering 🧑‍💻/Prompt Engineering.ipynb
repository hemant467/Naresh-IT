```python
import google.generativeai as genai
```

```python
from google.colab import userdata
api_key=userdata.get('GeminiKeyPrompt')
```

```python
genai.configure(api_key=api_key)
```

```python
prompt_template="""
                Context ; Role ; Constrains ; Instructions ; Examples ; Chain Of Thoughts

                Problem : Sentiment analysis
                Context : You are helping a customer service team understand customer feedback from reviews
                Role : You are a sentiment analysis AI assistant
                Constrains :
                - No Explanation
                - Review will be in single sentence
                - output must be exactly one word
                Instructions :
                - Read the input review and classify
                - Respond only with one word
                  - Positive
                  - Negative
                  - Neutral

                Examples
                User : The product was amazing and arrived on time
                Assistant : Positive

                User : This is the worst thing I ever bought
                Assistant : Negative

                User : It's okay, nothing special but not bad either
                Assistant : Neutral

                Now classify this below review
                User : I expect more from the product, but it works
                Assistant : Positive

                User: Totally love it, exceeded my expectations
                Assistant: Positive

                User: It broke after two days of use
                Assistant: Negative

                User: The packaging was fine, but I haven’t tried the product yet
                Assistant: Neutral

                User: Great value for the price!
                Assistant: Positive

                User: Didn’t meet the quality I was hoping for
                Assistant: Negative

                User: Not bad, not great, just average
                Assistant: Neutral

                User: Excellent customer service and fast delivery
                Assistant: Positive

"""
```

```python
genai.list_models()
```

```python
list(genai.list_models())
```

```python
model=genai.GenerativeModel('gemini-2.0-flash')
```

```python
# Model is ready with the variable of model
# Prompt is ready with the variable prompt_template

model.generate_content(prompt_template)
```

```python
response=model.generate_content(prompt_template)
```
