```python
def main():
    cap = cv2.VideoCapture(0) # 0 is the number of cameras
    draw = mp.solutions.drawing_utils # Helps in drawing the hand marks ; mp stands for Media Pipe

    try:
        while cap.isOpened():
            ret, frame = cap.read()

            if not ret:
                break
            frame = cv2.flip(frame, 1)
            frameRGB = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB) # Mediapipe requires the frame to be passed in RGB format for processing. OpenCV by default captures the frame in BGR format

            processed = hands.process(frameRGB) # To store the output i.e it contains all the Landmarks

            # Set-up variable to receive all the landmarks
            landmarks_list = []

            if processed.multi_hand_landmarks:
                hand_landmarks = processed.multi_hand_landmarks[0] # Taking the landmarks from 1 of the hand
                draw.draw_landmarks(frame, hand_landmarks, mpHands.HAND_CONNECTIONS) # Used to see / draw the landmarks

                # Loop all the Landmarks (lm) detector
                for lm in hand_landmarks.landmark:
                    landmarks_list.append((lm.x,lm.y)) # We are taking all the individual landmarks 1 by 1 & we'll push it into the landmarks_list ; lm.x, lm.y : Are the x,y Co-Ordinates of all the landmarks

                print(landmarks_list)

            cv2.imshow('Hemant Katta Frame',frame)
            if cv2.waitKey(1) & 0xFF == ord('q'):  # Wait for 1 milli second after each frame is read. If the keyboard input in the weight period is 'q' then break. Ord('q' will take the ASCII value of 'q'
                break

    finally:
        cap.release()
        cv2.destroyAllWindows()

if __name__ == '__main__':
    main()

# Each frame contains an array of 21 Landmarks i.e ( lm.x, lm.y : Are the x,y Co-Ordinates of all the landmarks )
```

```python
screen_width, screen_height = pyautogui.size()

mouse = Controller()
mpHands = mp.solutions.hands
hands = mpHands.Hands(
    static_image_mode=False, # We've set the static_image_mode to False since we're capturing video
    model_complexity=1, # To get better model
    min_detection_confidence=0.7, # Minimum confidence cover required for a hand recognition to be considered valid is 70%
    min_tracking_confidence=0.7, # Detect hand if it is more than 70% recognition
    max_num_hands=1 # Maximum number of hands to be detect is 1 hand
)
```

```python
def find_finger_tip(processed):
    if processed.multi_hand_landmarks:
        hand_landmarks = processed.multi_hand_landmarks[0] # Taking the landmarks from 1 of the hand
        return hand_landmarks.landmark[mpHands.HandLandmark.INDEX_FINGER_TIP]

    return None
```

```python
# Function for moving the mouse

# It gives the actual position co-ordinates of the x,y index finger on the screen

def move_mouse(index_finger_tip):
    if index_finger_tip is not None:
        x = int(index_finger_tip.x * screen_width)
        y = int(index_finger_tip.y * screen_height)
        pyautogui.moveTo(x,y) # Move the mouse using x,y co-ordinates of index finger tip
```

```python
def is_left_click(landmarks_list, thumb_index_dist):
    return (
            util.get_angle(landmarks_list[5],landmarks_list[6],landmarks_list[8]) < 50 and    # Checking index finger is bent / close
            util.get_angle(landmarks_list[9],landmarks_list[10],landmarks_list[12]) > 90 and  # Checking middle finger is straight
            thumb_index_dist > 50 # Checking thumb is straight
    )

def is_rigt_click(landmarks_list, thumb_index_dist):
    return (
            util.get_angle(landmarks_list[9],landmarks_list[10],landmarks_list[12]) < 50 and    # Checking middle finger is bent / close
            util.get_angle(landmarks_list[5],landmarks_list[6],landmarks_list[8]) > 90 and  # Checking index finger is straight
            thumb_index_dist > 50 # Checking thumb is straight
    )

def is_double_click(landmarks_list, thumb_index_dist):
    return (
            util.get_angle(landmarks_list[5],landmarks_list[6],landmarks_list[8]) < 50 and    # Checking index finger is bent / close
            util.get_angle(landmarks_list[9],landmarks_list[10],landmarks_list[12]) < 50 and  # Checking middle finger is straight
            thumb_index_dist > 50 # Checking thumb is straight
    )

# CHANGE
# Screenshot : If all the INDEX, MIDDLE & THUMB are closed
def is_screenshot(landmarks_list, thumb_index_dist):
    return (util.get_angle(landmarks_list[5],landmarks_list[6],landmarks_list[8]) < 50 and    # Checking index finger is bent / close
            util.get_angle(landmarks_list[9],landmarks_list[10],landmarks_list[12]) < 50 and  # Checking middle finger is bent / close
            thumb_index_dist < 50 # Checking thumb is bent / close
    )
```

```python
def detect_gestures(frame, landmarks_list, processed):
    if  len(landmarks_list)>=21: # 21 landmarks

        index_finger_tip = find_finger_tip(processed)
        thumb_index_dist = util.get_distance([landmarks_list[4],landmarks_list[5]]) # Distance between 4 & 5

        if thumb_index_dist < 50 and util.get_angle(landmarks_list[5],landmarks_list[6],,landmarks_list[8])>90: 
            # If the distance between the thumb & the index finger is less than 50 i.e it is closer to the base of the index finger
        # And if the index finger is upright position >90
            move_mouse(index_finger_tip)

        # LEFT click
        elif is_left_click(landmarks_list, thumb_index_dist):
            mouse.pree(Button.left)
            mouse.release(Button.left)
            cv2.putText(frame, "Left Click", (50,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)  # putText : Is used to put a text on the frame 50,50 pixel frame ; 255 color ; 2 thickness

        # RIGHT click
        elif is_right_click(landmarks_list, thumb_index_dist):
            mouse.pree(Button.right)
            mouse.release(Button.right)
            cv2.putText(frame, "Right Click", (50,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

        # DOUBLE click
        elif is_double_click(landmarks_list, thumb_index_dist):
            pyautogui.doubleClick()
            cv2.puText(frame, "Double Click", (50,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 0), 2)

        # SCREENSHOT
        elif is_screenshot(landmarks_list, thumb_index_dist):
            im1 = pyautogui.screenshot()  # im1 : image
            label = random.randint(1,1000) # Random numbers assigned to the Screenshot taken
            im1.save(f'my_screenshot_{label}.png')
            cv2.putText(frame,"Screenshot Taken", (50,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 0), 2)
        
        print(index_finger_tip) # Prints x,y co-Ordinates & z : Depth of the index finger i.e how far the index finger is from the camera
```

```python
# Utility Python file
# We're writing 2 functions 
# 1. Getting the Angle
# 2. Getting the Distance

import numpy as np

def get_angle(a,b,c): # We need 3 points a,b & c to calculate the angle between 2 lines i.e 2 fingers
    radians = np.arctan2(c[1] - b[1], c[0] - b[0]) - np.arctan2(a[1] - b[1], a[0] - b[0]) # Calculating angle between ab & bc in radians
    angle = np.abs(np.degrees(radians)) # Converting radians to degree ; abs - Absolute
    return angle

# Claculate the distance between 2 points
def get_distance(landmark_list):
    if len(landmark_list) < 2:  # If there are 2 landmarks we need to proceed further. Since we need 2 landmarks to calculate the Euclidean distance
        return

    # Take the (x,y) co-ordinates of the 2 points i.e landmarks
    (x1, y1),(x2, y2) = landmark_list[0], landmark_list[1] 
    L = np.hypot(x2 - x1, y2 - y1) # Calculate the Euclidean distance using the np.hypotenuse function c=sqrt[a^{2}+b^{2}]
    return np.interp(L, [0, 1], [0, 1000]) # Interpolation : The Euclidean distance value varies between [0,1] & multiplying to 1000
```
