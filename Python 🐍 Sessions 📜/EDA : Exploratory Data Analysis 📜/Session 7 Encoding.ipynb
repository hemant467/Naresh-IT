# LabelEncoder vs OneHotEncoder :

- Generally we dont want a relation between **input** columns

- We need both input columns are **NOT correlated** with each other

- We need both input columns are **independent(X,Y)** each other

- We need both input columns are **perpendicular** to each other

- We need both input columns are **Orthogonal** to each other

- We need both input columns are **90 degree phase shift** to each other

- **OneHotEncoder** will achieve above

- But OneHotEncoder creates new columns

  - Columns are increasing means dimensions are increasing
 
  - More dimensions data need more processing time
 
  - More dimensions data need more infra
 
  - This is called **Curse of Dimensionality**
