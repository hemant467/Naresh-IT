```python
continent_vc=visa_df['continent'].value_counts()
labels=continent_vc.keys()
count=continent_vc.values
continent_df=pd.DataFrame(zip(labels,count),columns=['Class','Frequency'])
continent_df
```

# Pie Charts :

```python
import os
folder_name='PieChart'
try:
    directory=os.makedirs(folder_name)
except Exception as e:
    print(e)
cwd=os.getcwd()
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(5,6)) # Value INCREASE
    plt.pie(values,labels=keys)
    file_name=f"{i}.jpg"
    path=os.path.join(cwd,folder_name,file_name)
    plt.savefig(path)
```
