```python
import matplotlib.pyplot as plt

n = len(cat)
plt.figure(figsize=(24, 10))

for i in range(1, n):
    plt.subplot(4, 2, i)
    
    col = cat[i]  # e.g., cat[1] = 'cont'
    keys = visa_df[col].value_counts().keys()
    values = visa_df[col].value_counts().values
    
    bars = plt.bar(keys, values, color=plt.cm.Set2.colors)  # Set2 has distinct pastel colors
    
    plt.title(f'Distribution of {col}',fontsize=14)
    plt.xticks(rotation=45, ha='right')
    
    # Add count on top of bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2, height, str(height),
                 ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```
