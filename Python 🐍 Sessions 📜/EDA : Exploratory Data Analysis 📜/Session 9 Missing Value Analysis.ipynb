# dropna :
- Drop the null value data, based on rows and columns

```python
# I want to extract the data without NULL VALUES
# axis = 0 means The Null Value will be removed based on Row
# axis = 1 means The Null Value will be removed based on Column

data1.dropna()
```

```python
data1.dropna(axis=1)
```

```python
dict2 = {'Name':['Akash','Rocky','Kushal',np.nan],
        'Age':[20,21,np.nan,23],
        'City':[np.nan,'Hyderabad','Hyderabad','Mumbai'],
        'Company':['Google','Microsoft','Apple','Meta']}
data2 = pd.DataFrame(dict2)
data2
```

```python
data2.dropna(axis=1)
```

# fillna() :

```python
data2.fillna(30)
```

# NOTE :
- The drawback is it add the value in every empty data location
- Even at **name** and **city**

```python
data2['Name'].fillna('Rohit')
```

- Drawback : Filling with Random value is not a Good Approch
- To avoid that we have some methods
    - backfill
    - bfill
    - ffill
    - pad

```python
import warnings
warnings.filterwarnings('ignore')
print(data2.fillna(method='pad',axis=1))
print()
print(data2.fillna(method='pad'))
```

```python
import warnings
warnings.filterwarnings('ignore')
print(data2.fillna(method='ffill',axis=1))
print()
print(data2.fillna(method='ffill'))
```

```python
import warnings
warnings.filterwarnings('ignore')
print(data2.fillna(method='backfill',axis=1))
print()
print(data2.fillna(method='backfill'))
```

```python
import warnings
warnings.filterwarnings('ignore')
print(data2.fillna(method='bfill',axis=1))
print()
print(data2.fillna(method='bfill'))
```

# Mean - Median - Mode :

- **Step 1 :** Select the **COLUMN**
- **Step 2 :** Get the **MEAN**
- **Step 3 :** Apply **fillna()**

```python
age_mean = round(data2['Age'].mean(),2)
print('age_mean :',age_mean)
data2['Age'].fillna(age_mean)
```

```python
city_mode = data2['City'].mode()

print('age_mode :',city_mode.values[0])
data2['City'].fillna(city_mode.values[0])
```

```python
from sklearn.impute import KNNImputer
x = [[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]]
pd.DataFrame(x)
```

```python
from sklearn.impute import KNNImputer
x = [[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]]
imputer = KNNImputer(n_neighbors=2,weights='uniform')
imputer.fit_transform(x)
```

```python
from sklearn.impute import KNNImputer
x = [[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]]
imputer = KNNImputer(n_neighbors=2,weights='distance')
imputer.fit_transform(x)
```
