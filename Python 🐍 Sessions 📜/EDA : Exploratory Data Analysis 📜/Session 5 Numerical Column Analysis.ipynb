```python
ids=['count','Min','p_25','Mean','Median','p_50','p_75','Max']
l=[]

for i in num:
    wage_data=visa_df[i]
    count=len(wage_data)
    Min=round(np.min(wage_data),2)
    p_25=round(np.percentile(wage_data,25),2) # 25 percent
    Mean=round(np.mean(wage_data),2)
    Median=round(np.median(wage_data),2)
    p_50=round(np.percentile(wage_data,50),2) # 50 percent
    p_75=round(np.percentile(wage_data,75),2) # 75 percent
    Max=round(np.max(wage_data),2)
    values=[count,Min,p_25,Mean,Median,p_50,p_75,Max]
    l.append(values)
```

```python
pd.DataFrame(l)
```

```python
pd.DataFrame(l,columns=ids,index=num)
```

```python
pd.DataFrame(l,columns=ids,index=num).T # Transpose
```

```python
pd.DataFrame(l,
             columns=ids, # Give index list at columns place
             index=num).T # Transpose, Give numerical values at index place
```
