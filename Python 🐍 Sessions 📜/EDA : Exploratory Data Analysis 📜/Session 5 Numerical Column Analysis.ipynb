# Emperical Rule : Normal Distribution

- u-1*std to u+1*std = 68%

- u-2*std to u+2*std = 95%

- u-3*std to u+3*std = 99.7%

- STEP - 1 : wage_mean

- STEP - 2 : wage_std

- STEP - 3 : lb=wage_mean-1*wage_std

- STEP - 4 : up=wage_mean+1*wage_std

- STEP - 5 : condition - 1 = wage_data > lb

- STEP - 6 : condition - 2 = wage_data < ub

- STEP - 7 : condition = condition - 1 & condition - 2

- STEP - 8 : len(visa_df[con]) == 68*25480/100

```python
wage_mean=np.mean(wage_data)
wage_mean
```

```python
wage_std = wage_data.std()
wage_std
```

```python
ub = wage_mean + 1 * wage_std
ub
```

```python
condition_1 = wage_data > lb
condition_2 = wage_data < ub

condition = condition_1 & condition_2
condition
```

```python
# STEP 8: Validate if approximately 68% of the data lies within 1 std deviation
expected_count = 68 * len(visa_df) / 100
actual_count = len(visa_df[condition])

print(f"Expected count (approx. 68%): {expected_count}")
print(f"\nActual count within 1 std deviation: {actual_count}")
```

# Task - 1 : Outlier both histogram and boxplot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
visa_df=pd.read_csv(r"E:\Python Projects\Data Files\Visadataset.csv")

cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

```python
cat
```

```python
num
```

```python
wage_data=visa_df['prevailing_wage']
wage_data
```

```python
q1 = np.percentile(wage_data,25)
q2 = np.percentile(wage_data,50)
q3 = np.percentile(wage_data,75)
IQR = q3 - q1
lb = q1 - 1.5 * IQR
ub = q3 + 1.5 * IQR
con1 = wage_data < lb
con2 = wage_data > ub
con = con1 | con2
outlier = wage_data[con]
len(outlier)
```

```python
plt.title('Outlier Histogram')
plt.hist(outlier,bins=40)
plt.show()
```

```python
plt.title('Outlier Box Plot')
plt.boxplot(outlier,vert=False)
plt.show()
```

# Task - 2 : Data without Outlier both histogram and boxplot

```python
q1 = np.percentile(wage_data,25)
q2 = np.percentile(wage_data,50)
q3 = np.percentile(wage_data,75)
IQR = q3 - q1
lb = q1 - 1.5 * IQR
ub = q3 + 1.5 * IQR
con1 = wage_data > lb
con2 = wage_data < ub
con = con1 & con2
non_outlier = wage_data[con]
len(outlier)
```

```python
plt.title('Data without Outlier Histogram')
plt.hist(non_outlier,bins=40)
plt.show()
```

```python
plt.title('Data without outlier box plot')
plt.boxplot(non_outlier,vert=False)
plt.show()
```

# Task - 3 : Replace outliers values with wage_median Histogram & Boxplot

```python
outliers_data=outlier.values
l = []
wage_data = visa_df['prevailing_wage'].values
wage_median = visa_df['prevailing_wage'].median()
for i in wage_data:
    if i in outliers_data:
        l.append(wage_median)
    else:
        l.append(i)

len(l)
```

```python
plt.title('Replace outlier values with wage_median Histogram')
plt.hist(l,bins=40)
plt.show()
```

```python
plt.title('Replace outlier values with wage_median Box Plot')
plt.boxplot(l,vert=False)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(25, 15))

# Function to set bold and dark axis labels and ticks
def style_axes():
    plt.title(plt.gca().get_title(), fontweight='bold', color='black')
    plt.tick_params(axis='both', colors='black', labelsize=10, labelcolor='black', direction='out', length=6, width=1)
    # Make tick labels bold
    for label in plt.gca().get_xticklabels() + plt.gca().get_yticklabels():
        label.set_fontweight('bold')
        label.set_color('black')

# Histograms with different colors
plt.subplot(4, 4, 1)
plt.hist(wage_data, bins=40, color='red')
plt.title('Data with Outliers - Histogram', fontweight='bold', color='black')
style_axes()

plt.subplot(4, 4, 2)
plt.hist(non_outlier, bins=40, color='orange')
plt.title('Data without Outliers - Histogram', fontweight='bold', color='black')
style_axes()

plt.subplot(4, 4, 3)
plt.hist(outlier, bins=40, color='blue')
plt.title('Outliers Only - Histogram', fontweight='bold', color='black')
style_axes()

plt.subplot(4, 4, 4)
plt.hist(l, bins=40, color='green')
plt.title('Outliers Replaced with Median - Histogram', fontweight='bold', color='black')
style_axes()

# Box plots with color and styled titles & ticks
def colored_boxplot(data, color, subplot_index, title):
    plt.subplot(4, 4, subplot_index)
    plt.boxplot(data, vert=False, patch_artist=True,
                boxprops=dict(facecolor=color),
                flierprops=dict(markerfacecolor='cyan', marker='o'),
                medianprops=dict(color='black'))
    plt.title(title, fontweight='bold', color='black')
    style_axes()

colored_boxplot(wage_data, 'coral', 5, 'Data with Outliers - Box Plot')
colored_boxplot(non_outlier, 'purple', 6, 'Data without Outliers - Box Plot')
colored_boxplot(outlier, 'teal', 7, 'Outliers Only - Box Plot')
colored_boxplot(l, 'maroon', 8, 'Outliers Replaced with Median - Box Plot')

plt.suptitle('Wage data comparison', fontsize=20, fontweight='bold', color='black', y=1.02)
plt.tight_layout()
plt.show()
```
