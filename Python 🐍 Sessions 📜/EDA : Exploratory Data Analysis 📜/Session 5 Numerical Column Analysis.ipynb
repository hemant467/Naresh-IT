# Emperical Rule : Normal Distribution

- u-1*std to u+1*std = 68%

- u-2*std to u+2*std = 95%

- u-3*std to u+3*std = 99.7%

- STEP - 1 : wage_mean

- STEP - 2 : wage_std

- STEP - 3 : lb=wage_mean-1*wage_std

- STEP - 4 : up=wage_mean+1*wage_std

- STEP - 5 : condition - 1 = wage_data > lb

- STEP - 6 : condition - 2 = wage_data < ub

- STEP - 7 : condition = condition - 1 & condition - 2

- STEP - 8 : len(visa_df[con]) == 68*25480/100

```python
wage_mean=np.mean(wage_data)
wage_mean
```

```python
wage_std = wage_data.std()
wage_std
```

```python
lb = wage_mean - 1 * wage_std
lb
```

```python
ub = wage_mean + 1 * wage_std
ub
```

```python
condition_1 = wage_data > lb
condition_2 = wage_data < ub

condition = condition_1 & condition_2
condition
```

```python
# STEP 8: Validate if approximately 68% of the data lies within 1 std deviation
expected_count = 68 * len(visa_df) / 100
actual_count = len(visa_df[condition])

print(f"Expected count (approx. 68%): {expected_count}")
print(f"\nActual count within 1 std deviation: {actual_count}")
```

```python
# STEP 8: Validate if approximately 68% of the data lies within 1 std deviation
len(visa_df[condition]) == 68*25480/100
```

```python
len(visa_df[condition]) == 95*25480/100
```

```python
len(visa_df[condition]) == 99.7*25480/100
```
