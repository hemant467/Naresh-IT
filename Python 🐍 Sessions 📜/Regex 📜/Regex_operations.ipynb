**re.match**

- Matches a pattern in a given string

```python
pattern=r"hello"
str1="hello world"

pattern in str1
# 'hello' in 'hello world' yes
```

```python
pattern=r"hello"
str1="hello world"
result=re.match(pattern,str1)
result.group()
```

```python
pattern=r"hello"
str1="world hello"
result=re.match(pattern,str1)
result
```

**Note:**

- re.match will checks for a match at beginning of string

- if we want to identify anywhere then we can go for **re.search**

**re.search**

- Searches pattern for anywhere in a given string

```python
pattern=r"hello"
str1="world hello"
result=re.search(pattern,str1)
print('result:',result)
result.group()
```

**re.findall**

- using a \d+ pattern

- \d indicates digit match between 0 to 9

- plus indicates one or more occurences of digit


```python
str1="the price is 100 and the tax is 20"
for i in str1.split():
    if i.isdigit():
        print(i)
```

```python
pattern=r"\d+"
str1="the price is 100 and the tax is 20"
result=re.findall(pattern,str1)
print('result:',result)
```

```python
str1="the price is 100 and the tax is 20"
l=str1.split()
for i in l:
    if i.isdigit():
        print(i)

# split
# for
# if
```

```python
pattern=r"\d+"
str1="order number 1234 shipped on 28th Decemeber 2024"
result=re.findall(pattern,str1)
print('result:',result)
```

**re.sub**

- substitute the string when pattern matchs

```python
pattern=r"\d+"
str1="order number 1234 shipped on 28th Decemeber 2024"
result=re.sub(pattern,'omkar',str1)
print('result:',result)
```

**Email validate use case**

- digits

- letters

- spl characters


```python
# omkar.nallagoni@gmail.com
# omkar.nallagoni
# @
# gmail
# ^ : used for strat of the string
# [a-zA-Z0-9_.+-]: pattern if any letter in email id have this 
#                  not only one character , many charcater 
#                  for that we need to add a + sign at last
# [a-zA-Z0-9_.+-]+
# till now the pattern is ^[a-zA-Z0-9_.+-]+

# @ : email id has symbol
# till now the pattern is ^[a-zA-Z0-9_.+-]+@

# [a-zA-Z0-9-]+
# till now the pattern is ^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+

# \.
# till now the pattern is ^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.

# com : [a-zA-Z]{2,}
# till now the pattern is ^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}

# $ : end of the string 

# final pattern is

pattern=r"^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}$"

# example
# user@example.com
# user.name+tag@sub-domain.example.org
# username@domain.co

# notmatching
# username@sub.domain
```

```python
pattern=r"^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}$"
email='omkar.nallagoni@gmail.com'
bool(re.match(pattern,email))
```

```python
pattern=r"^[a-zA-Z0-9]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}$"
email='omkar.nallagoni@gmail.com'
bool(re.match(pattern,email))
```

```python
pattern=r"^[a-zA-Z-]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}$"
email='omkar-nallagoni@gmail.com'
bool(re.match(pattern,email))
```
