```python
# STEP - 1 : xbar
arr_mean=np.mean(arr)
print(arr_mean)
# STEP - 2 : x-xbar
arr2=arr-arr_mean
print(arr2)
# STEP - 3 : (x-xbar)^2
arr3=np.square(arr2)
print(arr3)
# STEP - 1 : summation(x-xbar)
arr4=np.sum(arr3)
print(arr4)
var=arr4/len(arr)
print(var)
```

# **Distance between 2 points :**

```python
p1=(10,20) #(x1,y1)
p2=(5,6) #(x2,y2)
x1=p1[0]
y1=p1[1]
x2=p2[0]
y2=p2[1]
print(x1,y1)
print(x2,y2)
print(np.sqrt(np.square(x2-x1)+np.square(y2-y1)))
```

```python
np.array([10,20])-np.array([5,6])
```

```python
p1=(10,20)
p2=(5,6)
arr1=np.array(p1)
arr2=np.array(p2)
np.sqrt(np.sum(np.square(arr1-arr2)))
```

```python
p1=(10,20)
p2=(5,6)
arr1=np.array(p1)
arr2=np.array(p2)
round(np.sqrt(np.sum(np.square(arr1-arr2))),2)
```

# Linear Algebra :

```python
import numpy as np
p1=np.array((10,20))
p2=np.array((5,6))
round(np.linalg.norm(p1-p2),2)
```
