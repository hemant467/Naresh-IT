# Step-3: Chunk the data

- Sentence splitter

# Step-4: Convert into Embeddings

- save in a Fasis vector database

```python
import faiss
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.vector_stores.faiss import FaissVectorStore
from llama_index.core import VectorStoreIndex, StorageContext

# Use HF embedding model (you can replace with OpenAI or others)
embed_model = HuggingFaceEmbedding(model_name="sentence-transformers/all-MiniLM-L6-v2")

dummy_embedding = embed_model.get_text_embedding("test")
# we are passing a word test
# test is convert into vector
dimension = len(dummy_embedding)
faiss_index = faiss.IndexFlatL2(dimension)  # Flat index (no HNSW or IVFPQ)
vector_store = FaissVectorStore(faiss_index=faiss_index)
```

# Step-5: Store the vector index local

- StorageContext will give the idea how to store the embeddings

- How to retrive

- vector_store

- index_store

- metadata_store

```python
storage_context = StorageContext.from_defaults(vector_store=vector_store)
index = VectorStoreIndex(nodes,
                         storage_context=storage_context,
                         embed_model=embed_model)

# Persist index
storage_context.persist("storage/")
faiss.write_index(faiss_index, "storage/faiss_index.faiss")
```

# Step-6:Load the index

```python
import faiss
from llama_index.core import StorageContext, load_index_from_storage
from llama_index.vector_stores.faiss import FaissVectorStore

# Step 1: Load your FAISS index from the persisted location
faiss_index = faiss_index = faiss.read_index("storage/faiss_index.faiss")
  # or "storage/index.faiss" based on your filename
vector_store = FaissVectorStore(faiss_index=faiss_index)

# Step 2: Recreate the StorageContext with the vector store
storage_context = StorageContext.from_defaults(
    persist_dir="storage/",
    vector_store=vector_store
)

# Step 3: Load the index with the correct storage context
index = load_index_from_storage(storage_context, embed_model=embed_model)
```

```python
from llama_index.llms.groq import Groq
from llama_index.core.query_engine import RetrieverQueryEngine
from llama_index.core.response_synthesizers import get_response_synthesizer

# Step 1: Define the LLM
llm = Groq(model="llama3-8b-8192")
# 8b: 8 Billion parameters
# 8192: max tokens input+output


# Step 2: Get a response synthesizer with the LLM
response_synthesizer = get_response_synthesizer(llm=llm)

# Step 3: Create retriever
retriever = index.as_retriever(similarity_top_k=3)

# Step 4: Build the query engine
query_engine = RetrieverQueryEngine(
    retriever=retriever,
    response_synthesizer=response_synthesizer
)

# Step 5: Query
response = query_engine.query("Summarize the key takeaways")
print(response)
```
