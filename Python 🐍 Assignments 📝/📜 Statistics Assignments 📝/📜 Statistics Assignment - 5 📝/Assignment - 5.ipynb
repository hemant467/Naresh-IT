# 📊 Percentiles, Quartiles, and Box Plot :

## 🎯 1. Percentile

A **percentile** indicates the value below which a given percentage of observations fall.

### 📐 Definition:
The **p-th percentile** is the value below which **p%** of the data falls.

For example:
- 25th percentile (**P25**) → 25% of data is below this value
- 90th percentile (**P90**) → 90% of data is below this value

Percentiles help describe the **relative standing** of a value within a dataset.

---

## 🧮 2. Quartile

**Quartiles** are special percentiles that divide the data into **four equal parts**.

### 🔢 Quartiles:

- **Q1 (First Quartile)** = 25th percentile  
- **Q2 (Second Quartile)** = 50th percentile = **Median**  
- **Q3 (Third Quartile)** = 75th percentile  

They are used to measure **spread and center** of data.

### 📐 Interquartile Range (IQR):

**IQR = Q3 - Q1**

The IQR shows the **middle 50%** spread of the data, and is useful for identifying outliers.




---

## 📦 3. Box Plot (Box-and-Whisker Plot)

A **box plot** is a graphical representation of data that displays:

- **Minimum** (excluding outliers)
- **Q1 (25th percentile)**
- **Median (Q2 / 50th percentile)**
- **Q3 (75th percentile)**
- **Maximum** (excluding outliers)
- **Outliers** (plotted as individual points)




### 🖼️ Interpretation:
- The **box** shows the IQR (Q1 to Q3)
- The **line inside the box** is the median (Q2)
- **Whiskers** extend to the smallest and largest values within 1.5 × IQR
- **Points outside** that range are considered **outliers**

---

## 📈 Example :

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=100)

# Compute quartiles
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)  # or np.median(data)
q3 = np.percentile(data, 75)
iqr = q3 - q1

print("Q1:", q1, "Median:", q2, "Q3:", q3, "IQR:", iqr)

# Box plot
plt.boxplot(data, vert=False)
plt.title("Box Plot")
plt.xlabel("Value")
plt.show()


<h1><center><b>THE END</b></center></h1>
