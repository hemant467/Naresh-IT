# Covariance - Correlation :

## Covariance :

Covariance measures the direction of the relationship between two variables — whether they increase or decrease together.





- If co-variance has **POSITIVE** values, then the two variables are **POSITIVELY** correlated

- If co-variance has **NEGATIVE** values, then the two variables are **NEGATIVELY** correlated



# Interpretation:
- **Positive covariance** → variables tend to increase together

- **Negative covariance** → one increases as the other decreases

- **Zero covariance** → no linear relationship

- **Covariance** indicates direction, not strength.


## Correlation :

# Pearson Correlation Coefficient :


# Covariance vs Correlation :

| Feature           | Covariance                               | Correlation                                       |
| ----------------- | ---------------------------------------- | ------------------------------------------------- |
| **What it shows** | Direction of relationship                | Direction **and strength** of relationship        |
| **Scale**         | Not standardized (depends on units)      | Standardized (unitless, always between -1 and 1)  |
| **Range**         | **$-\infty$ to $+\infty$**               | **-1 to 1**                                       |
| **Units**         | Product of X and Y units                 | No units (dimensionless)                          |
| **Use case**      | Intermediate step to compute correlation | Compare strength of relationships across datasets |

# Uses :

- **Covariance :** When you only need to know the direction of movement, or as part of computing correlation, PCA, or covariance matrices.

- **Correlation :** When comparing relationships across different datasets, or interpreting strength & direction in a standardized way.

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Covariance matrix
cov_matrix = np.cov(x, y, ddof=1)
covariance = cov_matrix[0, 1]

# Correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]

print("Covariance:", covariance)
print("Correlation:", correlation)
```
