# 📊 Statistical Measures: CV, Skewness, and Kurtosis

These measures help us understand the **distribution and shape** of data beyond just mean and standard deviation.

---

## 1️⃣ Coefficient of Variation (CV)

### 📌 Definition:
The **Coefficient of Variation (CV)** is a **relative measure of dispersion**. It expresses the **standard deviation** as a **percentage of the mean**.

\[
\text{CV} = \frac{\sigma}{\mu} \times 100
\]

Where:
- \( \sigma \) = standard deviation  
- \( \mu \) = mean




### 🎯 Use Case:
- To compare variability **between datasets with different units or scales**
- Useful in fields like **finance**, **quality control**, **engineering**, and **biostatistics**

> ✅ Lower CV indicates **more consistency**, while higher CV suggests **greater variability**.

---

## 2️⃣ Skewness

### 📌 Definition:
**Skewness** measures the **asymmetry** of the data distribution.



- **Positive Skew (Right-skewed)**: Long tail on the right; mean > median  
- **Negative Skew (Left-skewed)**: Long tail on the left; mean < median  
- **Zero Skew**: Symmetrical distribution (like normal distribution)



\[
\text{Skewness} = \frac{1}{n} \sum_{i=1}^{n} \left( \frac{x_i - \bar{x}}{s} \right)^3
\]

Where:
- \( \bar{x} \) = sample mean  
- \( s \) = sample standard deviation

### 🎯 Use Case:
- Identify **distribution shape** and potential **outliers**
- Useful in **EDA**, **finance**, **risk management**, and **machine learning preprocessing**

---

## 3️⃣ Kurtosis

### 📌 Definition:
**Kurtosis** measures the **"tailedness"** or **peakedness** of a distribution.



Kurtosis} = \frac{1}{n} \sum_{i=1}^{n} \left( \frac{x_i - \bar{x}}{s} \right)^4

- **Mesokurtic (≈ 3)**: Normal distribution  
- **Leptokurtic (> 3)**: Heavy tails and sharp peak (more outliers)  
- **Platykurtic (< 3)**: Light tails and flatter peak (fewer outliers)

> Many software packages report **excess kurtosis**:  

**Excess Kurtosis = Kurtosis - 3**


### 🎯 Use Case:
- Detect **outliers**, **extreme events**, and **non-normality**
- Common in **finance**, **reliability engineering**, and **data modeling**

---

## 🧮 Python Example :

```python
import numpy as np
from scipy.stats import kurtosis, skew

data = np.random.normal(loc=50, scale=10, size=100)

# Coefficient of Variation
cv = np.std(data) / np.mean(data) * 100

# Skewness and Kurtosis
data_skewness = skew(data)
data_kurtosis = kurtosis(data)  # by default, excess kurtosis

print("CV (%):", cv)
print("Skewness:", data_skewness)
print("Excess Kurtosis:", data_kurtosis)

<h1><center><b>THE END</b></center></h1>
