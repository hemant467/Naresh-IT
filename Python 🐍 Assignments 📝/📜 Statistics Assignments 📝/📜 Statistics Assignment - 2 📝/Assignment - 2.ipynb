# Descriptive Statistics: Measures of Dispersion

**data = np.array([...])  # Replace [...] with data**

**n = len(data)**

**mean = np.mean(data)**



## 1. Range :
The **range** is the simplest measure of dispersion and is calculated as the difference between the maximum and minimum values in the data set.

**Range = max(x) - min(x)**

**range_ = np.max(data) - np.min(data)**

---

## 2. Mean Deviation :
The **mean deviation** is the average of the absolute differences between each data point and the mean of the data set.

**Mean Deviation} = [1 \ n] \ [sum_{i=1}^{n} |x_i - \bar{x}|]**

**mean_deviation = np.sum(np.abs(data - mean)) / n**

Where:
- x_i = each individual data point  
- bar{x} = mean of the data  
- n = total number of data points

---

## 3. Absolute Mean Deviation :
This is another name for the **mean deviation**. It emphasizes the use of absolute values in deviation calculation.

**absolute_mean_deviation = np.sum(np.abs(data - mean)) / n**

**Note:** **"Mean Deviation"** and **"Absolute Mean Deviation"** are often used interchangeably.

---

## 4. Variance :
**Variance** measures the average of the squared deviations from the mean. It gives more weight to extreme values.

variance = np.sum((data - mean)**2) / n

---

## 5. Standard Deviation :
The **standard deviation** is the square root of the variance and provides a measure of the average distance from the mean.

**std_dev = np.sqrt(variance)**



---

## 6.Sample Standar Deviation :
**sample_std_dev = np.sqrt(sample_variance)**

## 📋 Summary Table of Measures of Dispersion :

| 📌 Measure                  | 🧮 Formula / NumPy Function                                       | 🔍 Description                                       |
|----------------------------|------------------------------------------------------------------|------------------------------------------------------|
| **Range**                  | `np.max(data) - np.min(data)`                                   | Difference between max and min values                |
| **Mean Deviation**         | `np.mean(np.abs(data - np.mean(data)))`                         | Average of absolute deviations from the mean         |
| **Absolute Mean Deviation**| Same as Mean Deviation                                           | Another name for Mean Deviation                      |
| **Variance (Population)**  | `np.var(data)`                                                   | Average of squared deviations from the mean          |
| **Variance (Sample)**      | `np.var(data, ddof=1)`                                           | Sample variance with Bessel’s correction             |
| **Std Dev (Population)**   | `np.std(data)`                                                   | Square root of population variance                   |
| **Std Dev (Sample)**       | `np.std(data, ddof=1)`                                           | Square root of sample variance (with correction)     |
