# Understanding on Empirical rule - Chebyshev inequality :

# Empirical Rule (68–95–99.7 Rule) :

The Empirical Rule applies only to bell-shaped, normal distributions. It describes how much of the data falls within 1, 2, and 3 standard deviations from the mean.

# The Rule :
- **68%** of data falls within 1 standard deviation of the mean

- **95%** within 2 standard deviations

- **99.7%** within 3 standard deviations

# Formula :

If **𝜇** is the **mean** and **𝜎** is the **standard deviation**

- 𝜇 ± 1𝜎
    - **μ±1σ → ~68%**

- 𝜇 ± 2𝜎
    - **μ±2σ → ~95%**

- 𝜇 ± 3𝜎
    - **μ±3σ → ~99.7%**


The distribution is approximately **normal** or **symmetrical**



# Chebyshev’s Inequality :

**Chebyshev’s Inequality** applies to any distribution, regardless of shape. It gives a minimum proportion of values that lie within a certain number of standard deviations from the mean.

# Examples :

- For **𝑘 = 2** : At least **75%** of the data lies within **𝜇 ± 2𝜎**

- For **𝑘 = 3** : At least **88.89%** within **𝜇 ± 3𝜎**

- For **𝑘 = 4** : At least **93.75%**

# Used When :
- Distribution is not normal, or shape is unknown



## 📊 Comparison: Empirical Rule vs Chebyshev's Inequality :

| 🔍 Feature               | 📈 Empirical Rule                        | 📉 Chebyshev's Inequality                |
|-------------------------|------------------------------------------|------------------------------------------|
| **Applies to**          | Normal (bell-shaped) distributions       | All distributions (any shape)            |
| **1σ from mean**        | ~68% of data                             | ≥ 0% (no guarantee)                      |
| **2σ from mean**        | ~95% of data                             | ≥ 75% of data                            |
| **3σ from mean**        | ~99.7% of data                           | ≥ 88.89% of data                         |
| **4σ from mean**        | ~99.99% (very close to all)              | ≥ 93.75% of data                         |
| **Use Case**            | Quick estimates for normal data          | Guaranteed bounds for any distribution   |
| **Type of Estimate**    | Empirical (based on normal curve)        | Conservative (mathematically guaranteed) |
| **Strength**            | Precise but only for normal data         | General but less specific                |
