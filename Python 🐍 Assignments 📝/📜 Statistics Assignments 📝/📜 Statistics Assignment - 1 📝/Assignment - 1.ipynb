# 📊 Introduction to Statistics

---

## 1. What is Statistics?
Statistics is the science of collecting, analyzing, interpreting, and presenting data. It is broadly categorized into:

### ➤ Descriptive Statistics :
Describes and summarizes data using measures such as mean, median, mode, standard deviation, and visual tools like charts and graphs.

### ➤ Inferential Statistics :
Makes predictions or inferences about a population based on a sample of data. It involves confidence intervals, hypothesis testing, regression, etc.



---

## 2. Data Types of Variables

- **Qualitative (Categorical)**: 
  - Nominal (e.g., gender, color)
  - Ordinal (e.g., ranking, satisfaction level)

- **Quantitative (Numerical)**:
  - Discrete (e.g., number of children)
  - Continuous (e.g., height, weight)

---

## 3. Levels of Measurement

1. **Nominal** – Categories only (e.g., blood type)
2. **Ordinal** – Ordered categories (e.g., survey ratings)
3. **Interval** – Numerical with no true zero (e.g., temperature in °C)
4. **Ratio** – Numerical with a true zero (e.g., height, weight)



---

## 4. Population vs Sample

- **Population**: Entire group of individuals or items
- **Sample**: Subset of the population used to make inferences



---

## 13. Central Tendency

A measure that represents the center of a dataset:
- **Mean** (average)
- **Median** (middle value)
- **Mode** (most frequent value)







---

## 14. Mean vs Median

- **Mean** is affected by outliers
- **Median** is more robust to extreme values
- Use **median** when data is skewed or has outliers



---

## 15. Outliers

An outlier is a value significantly higher or lower than most other values in a dataset.

Can be detected using:
- IQR rule: values < Q1 – 1.5×IQR or > Q3 + 1.5×IQR
- Z-score method



---

## 16. Exact Meaning of Mode in Distributions

The **mode** is the value that appears most frequently in a dataset.

In distributions:
- **Unimodal**: one peak
- **Bimodal**: two peaks
- **Multimodal**: multiple peaks



---

## 17. Types of Distributions

- **Normal distribution**: Symmetric, bell-shaped curve
- **Skewed distribution**: Not symmetric
- **Uniform distribution**: All values have equal frequency

---

## 18. Positive Skew (Right Skew)

- Tail extends to the right
- Mean > Median > Mode

---

## 19. Negative Skew (Left Skew)

- Tail extends to the left
- Mean < Median < Mode

---

## 20. No Skew (Symmetric)

- Mean = Median = Mode
- Appears as a normal distribution
