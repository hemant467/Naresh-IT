```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings("ignore")

# Load data with correct delimiter
bank_df = pd.read_csv(r"E:\Python Projects\Data Files\bank.csv", delimiter=';')

# Separate categorical and numerical columns
cat = bank_df.select_dtypes(include='object').columns
num = bank_df.select_dtypes(exclude='object').columns

# Set up output directories
cwd = os.getcwd()
value_Counts_df_path = os.path.join(cwd, 'Value_Counts_df')
graphs_matplotlib_path = os.path.join(cwd, 'Graphs_matplotlib')
graphs_seaborn_path = os.path.join(cwd, 'Graphs_Seaborn')

# Create directories if they don't exist
for path in [value_Counts_df_path, graphs_matplotlib_path, graphs_seaborn_path]:
    os.makedirs(path, exist_ok=True)

# Save Value Counts to CSVs
for i in range(len(cat)):
    keys = bank_df[cat[i]].value_counts().keys()
    values = bank_df[cat[i]].value_counts().values
    df = pd.DataFrame(values, index=keys, columns=['Count'])
    file_name = f'{i}_{cat[i]}.csv'
    df.to_csv(os.path.join(value_Counts_df_path, file_name))

# Multicolor Bar Charts (Matplotlib)
for i in range(len(cat)):
    keys = bank_df[cat[i]].value_counts().index
    values = bank_df[cat[i]].value_counts().values

    plt.figure(figsize=(17, 5))
    plt.title(f'Bar Chart - {cat[i]}')

    # Use Set3 for distinct pastel colors
    colors = sns.color_palette("Set3", n_colors=len(keys))
    plt.bar(keys, values, color=colors)

    plt.xticks(rotation=45)
    plt.tight_layout()

    file_name = f'{i}_bar_{cat[i]}.png'
    plt.savefig(os.path.join(graphs_matplotlib_path, file_name))
    plt.close()

# Improved Multicolor Pie Charts (Matplotlib)
for i in range(len(cat)):
    keys = bank_df[cat[i]].value_counts().index
    values = bank_df[cat[i]].value_counts().values

    plt.figure(figsize=(17, 6))
    plt.title(f'Pie Chart - {cat[i]}')

    num_labels = len(keys)
    colors = sns.color_palette("Set3", n_colors=num_labels)
    explode = [0.02] * num_labels  # slight separation for each slice

    plt.pie(
        values,
        labels=keys,
        autopct='%0.2f%%',
        colors=colors,
        explode=explode,
        startangle=140,
        wedgeprops={'edgecolor': 'black'}
    )

    plt.tight_layout()
    file_name = f'{i}_pie_{cat[i]}.png'
    plt.savefig(os.path.join(graphs_matplotlib_path, file_name))
    plt.close()

# Multicolor Seaborn Countplots
for i in range(len(cat)):
    plt.figure(figsize=(19, 5))
    
    unique_vals = bank_df[cat[i]].nunique()
    palette = sns.color_palette("Set2", n_colors=unique_vals)

    sns.set(style="ticks", context="talk")
    sns.countplot(data=bank_df, x=cat[i], palette=palette)

    plt.title(f'Seaborn Countplot - {cat[i]}')
    plt.xticks(rotation=45)
    plt.tight_layout()

    file_name = f'{i}_{cat[i]}.png'
    plt.savefig(os.path.join(graphs_seaborn_path, file_name))
```
