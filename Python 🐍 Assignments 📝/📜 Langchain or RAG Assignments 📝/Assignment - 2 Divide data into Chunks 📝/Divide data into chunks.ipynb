# Text Loader :

````python
from langchain.document_loaders import TextLoader

# Use a raw string (prefix with r) or double backslashes to avoid escape issues
file_path = r"E:\Python Projects\Data Files\State_union.txt"

# Explicitly specify encoding='utf-8' to avoid UnicodeDecodeError
loader = TextLoader(file_path, encoding='utf-8')

# Load the data
txt_data = loader.load()
print(txt_data[0])
````

# Dividing data into Chunks :

````python
from langchain.document_loaders import PyPDFLoader

# Path to your converted PDF file
pdf_path = r"E:\Python Projects\Data Files\State_union_converted.pdf"

# Create the PDF loader
loader = PyPDFLoader(pdf_path)

# Load the documents (returns a list of Document objects)
pdf_data = loader.load()

# Loop through each document chunk and print its content
for i, doc in enumerate(pdf_data):
    print(f"--- Page chunk {i+1} ---")
    print(doc.page_content)
    print("\n")  # Add extra newline for readability
````

# Code Explanation and Observations

This code demonstrates how to load a PDF file using LangChain’s **`PyPDFLoader`** and print its content in chunks.

---

### Step-by-step breakdown:

1. **Import the PDF loader :**  
   ```python
   from langchain.document_loaders import PyPDFLoader

- This imports the PyPDFLoader class, which is designed to read and parse PDF files.

2. **Specify the PDF file path :**
    pdf_path = r"E:\Python Projects\Data Files\State_union_converted.pdf"

- The raw string (prefixed with r) ensures that backslashes in the Windows file path are handled correctly, avoiding escape sequence issues.

3. **Create a loader instance :**
    loader = PyPDFLoader(pdf_path)

- This initializes the loader with the given PDF file, preparing it for reading.

4. **Load the PDF content :**
    pdf_data = loader.load()

- The load() method reads the PDF and returns a list of Document objects. Each object typically represents a chunk of text (usually one page or a section).

5. **Iterate and print each chunk :**
    for i, doc in enumerate(pdf_data):
    print(f"--- Page chunk {i+1} ---")
    print(doc.page_content)
    print("\n")

- This loop goes through each chunk and prints its index and text content. It helps verify the extraction and allows you to inspect the content page-by-page or chunk-by-chunk.

# Summary Observations:
