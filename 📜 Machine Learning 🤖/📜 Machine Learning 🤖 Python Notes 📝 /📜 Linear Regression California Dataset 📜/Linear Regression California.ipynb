## Step-3: Feature selection: VIF


- VIF is a concept related to two variables are multicollinear or not

- Every time one variable behave as dependent and remaining are independent variable

- It will give R_Square means Explainability 

- More R_square means more relation



```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Compute the correlation matrix
corr_matrix = scaled_df.corr()
print(corr_matrix)

# Visualize the correlation matrix
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
X = scaled_df.drop('MedHouseVal', axis=1) # drop the output column
vif_list=[]
for i in range(len(X.columns)):
    vif_list.append(variance_inflation_factor(X.values,i))
vif_data=pd.DataFrame(vif_list,index=X.columns,columns=['VIF'])
vif_data
```

```python
# seperate the dataframe two parts
# VIF>5  VIF<5
# VIF<5 ==== all we need  : 4 columns
# VIF>5 === 4 columns AveRooms,AveBedrms,Latitude,Longitude
# AveRooms,AveBedrms  both are mulitcolinear each other
#     we need to pick any one column
# Latitude,Longitude both are mulitcolinear each other
#     we need to pick any one column

# Final data 6 input + 1target 
```

<img src="https://miro.medium.com/v2/resize:fit:1400/0*qNXzav-2ksKGws64.png" jsaction="" class="sFlh5c FyHeAf iPVvYb" style="max-width: 1400px; height: 61px; margin: 42.5px 0px; width: 305px;" alt="ML Series 5: Understanding R-squared in Regression Analysis | by Sahin  Ahmed, Data Scientist | Medium" jsname="kn3ccd">

```python
y_actual=df['Y_actual']
y_pred=df['Y_predict']
############ error #################
error=y_actual-y_pred
#=============Square error ======================
se=np.square(error)
#================ SSE ================================
sse=np.sum(se)
#================= MSE ===============================
mse=sse/len(X_test)
#=====================RMSE=========================
rmse=np.sqrt(mse)
#=================== RSE ===========================
p=len(X_test.columns)
n=len(X_test)
rse=np.sqrt(sse/(n-p-1))
#================= Rsquare =========================
y_avg=np.mean(y_actual)
dr=np.sum(np.square(y_test-y_avg))
r2_score=1-(sse/dr)

print('SSE:',sse)
print('MSE:',mse)
print('RMSE:',rmse)
print('RSE:',rse)
print('R_Square:',r2_score)
```

**Real time vs Batch prediction**

- model will take one tuple of column data only

- we trained by using 6 columns

- if we want a prediction make sure pass all 6 columns data at a time

- sending one tuple or one list it is called as **Real time predictions**

- if we send multiple tuples or lists , it is called as **Batch prediction**

```python
i1=[20,30,40,50,60,70]
lr.predict([i1])
```
